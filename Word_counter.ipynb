```python
%%writefile counter/views.py

from django.shortcuts import render

def home(request):
    text = ""
    word_count = 0
    character_count = 0
    characters_without_spaces = 0

    if request.method == "POST":
        text = request.POST.get("text", "")
        word_count = len(text.split())
        character_count = len(text)
        characters_without_spaces = len(text.replace(" ", ""))

    return render(request, "counter/home.html", {
        "text": text,
        "word_count": word_count,
        "character_count": character_count,
        "characters_without_spaces": characters_without_spaces,
    })
```

```python
%%writefile counter/urls.py

from django.urls import path
from . import views

urlpatterns = [
    path("", views.home, name="home"),
]
```

```python
%%writefile README.md

# Word Counter - Django

## Project Description

A simple web application developed using Python and Django.
It allows users to enter text and calculate the number of words
and characters.

## Features

- Count total words
- Count total characters
- Count characters without spaces
- Clear the text area
- Simple and user-friendly interface

## Technologies Used

- Python
- Django
- HTML
- CSS
- SQLite

## How to Run

1. Install the required packages:

   pip install -r requirements.txt

2. Apply migrations:

   python manage.py migrate

3. Start the server:

   python manage.py runserver

4. Open the application in a browser.

## How It Works

The user enters text into the webpage. Django receives the text
through a POST request, processes it, calculates the word and
character counts, and displays the results.

## Project Structure

- manage.py — Django project manager
- counter — Main application
- views.py — Word counting logic
- urls.py — URL routing
- home.html — User interface
- settings.py — Django configuration

## Author

Django Academic Project
```
